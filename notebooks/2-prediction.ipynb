# Classify a small area

Runs the same functions as `mrd run` on a small window and shows the result. Nothing is written to S3.

```python
from dep_tools.grids import PACIFIC_GRID_10
from matplotlib.colors import ListedColormap

from mrd import CLASSES, MODEL_URL
from mrd.data import load_data, load_model, predict
```

## Features

A window of tile 65,21 near Taveuni's west coast; any `PACIFIC_GRID_10.tile_geobox((col, row))` (or slice of one) works.

```python
window = PACIFIC_GRID_10.tile_geobox((65, 21))[1300:1556, 8500:9012]
ds = load_data(window, 2022)
ds
```

## Model

The published model for this version (needs `mrd train` to have uploaded it), or a local one such as
`../data/models/0-0-2/model.joblib` from `mrd train --no-upload`.

```python
model = load_model(MODEL_URL)
out = predict(model, ds).compute()
out
```

## Map

```python
colours = ListedColormap([colour for _, colour in CLASSES.values()])
out["class"].where(out["class"] != 255).odc.explore(cmap=colours, vmin=min(CLASSES) - 0.5, vmax=max(CLASSES) + 0.5)
```
