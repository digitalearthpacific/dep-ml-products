```python
import pystac
from odc.stac import load

item = pystac.Item.from_file(
    "https://dep-public-staging.s3.us-west-2.amazonaws.com/dep_s2s1_mrd/0-0-2/064/020/2022/dep_s2s1_mrd_064_020_2022.stac-item.json",
)
ds = load([item], chunks={})
ds
```

```python
from matplotlib.colors import ListedColormap

from mrd import CLASSES, NODATA

# One colour per class code 1-8, in order; vmin/vmax centre each code on its colour.
cmap = ListedColormap([colour for _, colour in CLASSES.values()])
classes = ds["class"].squeeze("time")
classes.where(classes != NODATA).odc.explore(cmap=cmap, vmin=0.5, vmax=len(CLASSES) + 0.5)
```

```python
import branca.colormap as cm

min = 0
max = 100
m = ds["proba"].where(classes != NODATA).odc.explore(vmin=min, vmax=max, cmap="RdYlGn")
cm.linear.RdYlGn_11.scale(min, max).add_to(m)
m
```
