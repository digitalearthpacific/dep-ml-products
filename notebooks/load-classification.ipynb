```python
import pystac
import json
from odc.stac import load

from mrd.classify import print_tasks

tasks = json.loads(print_tasks(years="2022", overwrite=True))

items = []
for task in tasks:
    print(task)
    x, y = task["tile-id"].split(",")
    x = f"{int(x):03}"
    y = f"{int(y):03}"
    year = task["year"]

    url = f"https://dep-public-staging.s3.us-west-2.amazonaws.com/dep_s2s1_mrd/{task['version']}/{x}/{y}/{year}/dep_s2s1_mrd_{x}_{y}_{year}.stac-item.json"
    try:
        items.append(pystac.Item.from_file(url))
    except Exception as e:
        print(f"Failed to load {url}: {e}")

ds = load(items, chunks={}, bands=["class"]) # "proba" band not loaded
ds
```

```python
from matplotlib.colors import ListedColormap

from mrd import CLASSES, NODATA

# One colour per class code 1-8, in order; vmin/vmax centre each code on its colour.
cmap = ListedColormap([colour for _, colour in CLASSES.values()])
classes = ds["class"].squeeze("time")
classes.where(classes != NODATA).odc.explore(cmap=cmap, vmin=0.5, vmax=len(CLASSES) + 0.5)
```
